```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return(A * np.sin(w*x))
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
linestyles = ['-', '--', '-.', ':', (0, (3, 1, 1, 1))]
```

```python
fig, axes = plt.subplots(1, 3, figsize = (25, 5))
for A, w, ls in zip(amplitude, frequency, linestyles):
    y1 = make_sine_wave(x, A, 1)
    axes[0].plot(x, y1, linestyle = ls, linewidth = 2, label = f'A = {A}, w = 1')
    y2 = make_sine_wave(x, 1, w)
    axes[1].plot(x, y2, linestyle = ls, linewidth = 2, label = f'A = 1, w = {w}')
    y3 = make_sine_wave(x, A, w)
    axes[2].plot(x, y3, linestyle = ls, linewidth = 2, label = f'A = {A}, w = {w}')

titles = [
    'Varied amplitude',
    'Varied frequency',
    'Varied amplitude and frequency'
]
for ax, title in zip(axes, titles):
    ax.set_title(title, fontsize = 15)
    ax.set_xlabel('x (radians)', fontsize = 12)
    ax.set_ylabel('y = A * sin(wx)', fontsize = 12)
    ax.legend(fontsize = 10, loc = 'upper right')

plt.show()
```
