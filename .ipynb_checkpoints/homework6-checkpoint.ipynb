## Prime Star Systems

```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def prime_star(arr):
    def is_prime(n):
        if n < 2:
            return False
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                return False
            return True
    prime_check = np.vectorize(is_prime)
    prime_apply = prime_check(arr)
    return arr[np.any(prime_apply, axis = 1)]

result = prime_star(arr)
print(result)
```

## 5. Let's Play Checkers!

#### 5.1, 5.2, 5.3

```python
def checkerboard():
    board = np.zeros((8, 8), dtype = int)

    board[::2, ::2] = 1

    board[1::2, 1::2] = 1

    return board

print(checkerboard())
```

#### 5.4

```python
def reversed_checkerboard():
    board = np.zeros((8, 8), dtype = int)

    board[::2, 1::2] = 1

    board[1::2, ::2] = 1

    return board

print(reversed_checkerboard())
```

## 6. The Expanding Universe

```python
universe = np.array(["supernovae", "pulsars", "galaxies"])

def expansion(universe, num_spaces):
    spacer = " " * num_spaces
    return np.char.join(spacer, universe)

print(expansion(universe[1], 1))
print(expansion(universe[2], 1))
```

## 7. Second-Brightest Star

```python
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
```

```python
def second_brightest(stars):
    return np.sort(stars, axis = 1)[:, -2]

print(second_brightest(stars))
```
