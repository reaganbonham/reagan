##### I chose the planets.csv file

```python
dF_full = sns.load_dataset('planets')
print(dF_full)
```

```python
dF = dF_full.dropna(axis=1)
print(dF)
```

##### The method describes how the planet was detected (raidal velocity or transit), the number is how many planets, and the year is when it was detected

```python
fig, axes = plt.subplots(2, 2, figsize = (45, 20))

sns.scatterplot(data=dF, x='method', y='number', color='olivedrab', hue='method', ax=axes[0, 0])
axes[0, 0].set_title('method vs. number')

sns.scatterplot(data=dF, x='number', y='method', color='darkolivegreen', ax=axes[0, 1])
axes[0, 1].set_title('number vs. method')

sns.histplot(data=dF, x='year', color='darkgreen', ax=axes[1, 0])
axes[1, 0].set_title('year of discovering planets')

sns.boxplot(data=dF, x='method', color='forestgreen', ax=axes[1, 1])
axes[1, 1].set_title('method of discovering planets')

plt.show()
```

##### The scatterplots represent the number of planets discovered versus the method in which they were discovered and vice versa; so essentially the two are just flipped.
##### The histogram represents the number of planets discovered in each year. Overwhelmingly, 2010 appears to be the year in which the most planets were discovered
##### The boxplot is just the method in which each planet was discovered. Most were radial velocity to transit
